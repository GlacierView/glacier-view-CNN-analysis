# 08 — Inventory what has been downloaded

**Purpose.** Records what the time-series pull (notebook 05) has put on disk: every `(glims_id,
file_name)` pair under the landing zone. Useful for checking progress, resuming a partial pull, or
comparing local files against S3. Nothing else in the repo reads the CSV.

## Inputs
- **`config.landsat_dir(config.INFERENCE_DATA_LABEL)`**: per-glacier folders from notebook 05.

## Outputs
- `already_pulled_ee_files.csv` (`glims_id`, `file_name`) in
  `config.processed_metadata_dir(config.INFERENCE_DATA_LABEL)`. Scenes and metadata files, not
  directories.

```python
import pandas as pd

from glacierview import config

LABEL = config.INFERENCE_DATA_LABEL
pulled = pd.DataFrame(
    [(p.relative_to(config.landsat_dir(LABEL)).parts[0], p.name)
     for p in sorted(config.landsat_dir(LABEL).rglob("*")) if p.is_file()],
    columns=["glims_id", "file_name"],
)
pulled.to_csv(config.processed_metadata_dir(LABEL) / "already_pulled_ee_files.csv", index=False)
pulled.groupby("glims_id").size().describe()
```
